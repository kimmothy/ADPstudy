```python
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

iris_raw = load_iris()
iris = pd.DataFrame(iris_raw.data, columns=iris_raw.feature_names)
iris['class'] = iris_raw.target
iris['class'] = iris['class'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

del iris_raw


```

```python
def linear_func(x):
    return 2 * x + 1

X = iris["sepal_length"]

#이건 Y값이 아니라, x값이야. y값은 linear_func(X)로 구할 수 있어.
plt.plot(X, linear_func(X), label='y=2x+1', c="#789345")
plt.show()
```

```python
import numpy as np

X, Y = iris["sepal_length"], iris["petal_length"]
b1, b0 = np.polyfit(X, Y, 1)

plt.scatter(x = X, y = Y, alpha=0.5)

plt.plot(X, b1*X + b0, color="red", marker="d")

plt.show()
```

```python
iris2 = iris.sort_values(by="sepal_length", ascending=True)
X, Y = iris2["sepal_length"], iris2["petal_length"]
b2, b1, b0 = np.polyfit(X, Y, 2)
plt.scatter(x = X, y = Y, alpha=0.5)
plt.plot(X, b2*X**2 + b1*X + b0, color="red", marker="o")
plt.show()
```

```python
plt.plot('sepal_length', 'petal_length', data=iris2, marker='o', color='red', label='sepal_length vs petal_length')
plt.show()
```

```python
plt.plot('sepal_length', 'petal_length', data=iris2.loc[iris2["class"] == "setosa"], marker='o', color='green')
plt.plot('sepal_length', 'petal_length', data=iris2.loc[iris2["class"] == "versicolor"], marker='o', color='yellow')
plt.plot('sepal_length', 'petal_length', data=iris2.loc[iris2["class"] == "virginica"], marker='o', color='blue')
# plt.label('sepal_length vs petal_length by class')

plt.show()
```
