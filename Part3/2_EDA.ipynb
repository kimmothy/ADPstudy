```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
```

```python
wine_raw_data = load_wine()
wine_df = pd.DataFrame(wine_raw_data.data, columns=wine_raw_data.feature_names)
wine_df["class"] = wine_raw_data.target
wine_df["class"] = wine_df["class"].map({0: "class_0", 1: "class_1", 2: "class_2"})
print(wine_df.head())
```

```python
wine_type = wine_df["class"].value_counts()
wine_type
```

```python
plt.bar(wine_type.index, wine_type.values, width=0.5, bottom=None, align="center")
```

```python
plt.barh(wine_type.index, wine_type.values, height=0.4, left=None, align="edge")
plt.show()
```
