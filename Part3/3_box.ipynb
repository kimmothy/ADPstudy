```python
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

iris_raw = load_iris()
iris = pd.DataFrame(iris_raw.data, columns=iris_raw.feature_names)
iris['class'] = iris_raw.target
iris['class'] = iris['class'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

plt.boxplot(iris.drop(columns='class'))
plt.show()
```

```python
plt.boxplot(iris['sepal_width'],whis=1.5)
plt.title("Sepal Width by CM")
plt.show()
```

```python
iris[['class', 'sepal_width']].boxplot(by='class')
plt.show()
```

```python
import seaborn as sns
sns.boxplot(x='class', y='sepal_width', data=iris)
plt.show()
```
