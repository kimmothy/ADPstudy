```python
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt

iris_raw = load_iris()
iris = pd.DataFrame(iris_raw.data, columns=iris_raw.feature_names)
iris['class'] = iris_raw.target
iris['class'] = iris['class'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

del iris_raw

plt.title("iris scatter")
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")

# plt.scatter(x=iris['sepal_length'], y=iris['sepal_width'])
plt.scatter(x=iris['sepal_length'], y=iris['sepal_width'], c=iris['class'].map({'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}))
```

```python
import seaborn as sns

sns.scatterplot(x='sepal_length', y='sepal_width', hue='class', data=iris, style='class')

plt.show()
```
