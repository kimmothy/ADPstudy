## 핵심 요약
train_test_split은 잘 나눠준다

```python
# 데이터 로드 부분

import pandas as pd
from sklearn.datasets import load_iris

iris_raw = load_iris()
iris = pd.DataFrame(data=iris_raw.data, columns=iris_raw.feature_names)
iris['class'] = iris_raw.target
iris['class'] = iris['class'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris.drop('class', axis=1), iris['class'], test_size=0.2, random_state=1004)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(iris.drop('class', axis=1), iris['class'], test_size=0.2, stratify=iris['class'])
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
y_train.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

stdScaler = StandardScaler()

# transform() 메서드는 fit() 메서드로 계산된 평균과 표준편차를 사용하여 데이터를 변환합니다.
#X_train_scaled = stdScaler.transform(X_train)
# fit_transform() 메서드는 fit()과 transform()을 한 번에 수행하는 메서드입니다. 즉, 데이터를 표준화하기 위해 평균과 표준편차를 계산하고, 그 계산된 값을 사용하여 데이터를 변환합니다.
# 둘 중 하나를 사용하면 되지만, fit_transform() 메서드를 사용하는 것이 더 간편합니다.
X_train_scaled = stdScaler.fit_transform(X_train)

X_test_scaled = stdScaler.transform(X_test)
```

```python
print("\t\t(min, max) (mean, std)")
print("Train_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_train_scaled.min(), X_train_scaled.max(), X_train_scaled.mean(), X_train_scaled.std()))
print("Test_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_test_scaled.min(), X_test_scaled.max(), X_test_scaled.mean(), X_test_scaled.std()))
```

```python
from sklearn.preprocessing import MinMaxScaler

minMaxScaler = MinMaxScaler()

X_train_scaled = minMaxScaler.fit_transform(X_train)
X_test_scaled = minMaxScaler.transform(X_test)
print("\t\t(min, max) (mean, std)")
print("Train_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_train_scaled.min(), X_train_scaled.max(), X_train_scaled.mean(), X_train_scaled.std()))
print("Test_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_test_scaled.min(), X_test_scaled.max(), X_test_scaled.mean(), X_test_scaled.std()))
```

```python
from sklearn.preprocessing import MaxAbsScaler

maxabsScaler = MaxAbsScaler()

X_train_scaled = maxabsScaler.fit_transform(X_train)
X_test_scaled = maxabsScaler.transform(X_test)
print("\t\t(min, max) (mean, std)")
print("Train_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_train_scaled.min(), X_train_scaled.max(), X_train_scaled.mean(), X_train_scaled.std()))
print("Test_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_test_scaled.min(), X_test_scaled.max(), X_test_scaled.mean(), X_test_scaled.std()))
```

```python
from sklearn.preprocessing import RobustScaler

robScaler = RobustScaler()

X_train_scaled = robScaler.fit_transform(X_train)
X_test_scaled = robScaler.transform(X_test)
print("\t\t(min, max) (mean, std)")
print("Train_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_train_scaled.min(), X_train_scaled.max(), X_train_scaled.mean(), X_train_scaled.std()))
print("Test_scaled (%.2f, %.2f) (%.2f, %.2f)" % (X_test_scaled.min(), X_test_scaled.max(), X_test_scaled.mean(), X_test_scaled.std()))
```
