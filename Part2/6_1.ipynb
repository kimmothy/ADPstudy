```python
import pandas as pd
HR1 = pd.DataFrame({'이름': ["장화", "홍련"], '부서':["영업","회계"], '직급':["팀장","사원"]})
HR2 = pd.DataFrame({'이름': ["콩쥐", "팥쥐"], '직급':["선임","책임"], '부서':["영업","구축"]})
print(pd.concat([HR1, HR2]))
```

```python
print(pd.concat([HR1, HR2], axis=0, ignore_index=True))
```

```python
HR3 = pd.DataFrame({'이름': ["콩쥐", "팥쥐"], '부서':["영업","구축"], "급여": [300, 400]})
print(pd.concat([HR1, HR3], axis=0, ignore_index=True))
```

```python
HR4 = pd.Series([300, 500], name="급여")
print(pd.concat([HR1, HR3], axis=1))
```
