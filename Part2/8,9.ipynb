```python
import pandas as pd
scores = {"학년": [1,1,1,1,2,2],
          '반': ["A","A","B","B","C","C"],
          "성별": ["남","여","남","여","남","여"],
          "점수": [90, 85, 80, 95, 88, 92]}
df = pd.DataFrame(scores)
df
```

```python
df = df.pivot_table(index=["학년", "반"], columns="성별", values="점수")
df
```

```python
df.reset_index()
```

```python
df.reset_index().melt(id_vars=["학년", "반"], var_name="성별", value_name="점수")
```

```python
import numpy as np

df.apply(np.sqrt, axis=0)
```

```python
df.apply(np.max, axis=0)
```

```python
df.apply(np.max, axis=1)
```

```python
def plus_five(val):
    return val + 5
df.apply(plus_five, axis=0)
```

```python
def class_avg(df):
    return int((df["남"] + df["여"]) / 2)

df.apply(class_avg, axis=1)
```
