```python
score = pd.DataFrame({"국어": [90, 80, 70], "영어": [85, 75, 65], "수학": [95, 85, 75]}, index=['홍길동', '김철수', '이영희'])
score.loc['홍길동', '영어'] = 100
score['국어'] = score['국어'] - 3
score
```

```python
new_students = pd.DataFrame({"국어": [88, 78], "영어": [82, 72], "수학": [92, 82]}, index=['박영수', '최민수'])
score = pd.concat([score, new_students])
# score = score.append(new_students) # append는 pandas 1.4.0 이후로 deprecated 되었음
score
```

```python
another_student1 = pd.DataFrame({"국어": [85], "영어": [80], "수학": [90]}, index=['김영수'])
score = pd.concat([score, another_student1])
score
```

```python
science_score =[85, 75, 65, 80, 70, 20]
score['과학'] = science_score
score['학년'] = 1
score
```

```python
social_score = pd.DataFrame({"사회": [80, 70, 60, 75, 65, 30]}, index=score.index)
score = pd.concat([score, social_score], axis=1)
score
```

```python
magic_score = pd.Series([90, 80, 70, 85, 75, 25], index=score.index)
score['마법'] = magic_score
score
```

```python
score['총점'] = score.sum(axis=1) -1
score
```

```python
score.drop('홍길동', inplace=True)
score.drop('마법', inplace=True, axis=1)
score
```
