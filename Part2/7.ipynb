```python
import pandas as pd
from sklearn.datasets import load_iris

IRIS = load_iris()
iris= pd.DataFrame(IRIS.data, columns=IRIS.feature_names)

iris.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

iris['class']  = IRIS.target
iris['class']  = iris['class'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})
```

```python
print(IRIS.target)
print(type(IRIS.target))
iris.columns
```

```python
iris.groupby('class').median()
```

```python
print(pd.Series(iris['class']).value_counts())
print(iris['class'].value_counts())
print(type(iris['class']))
print(type(iris['class'].value_counts()))
```

```python
iris['petal_width_level'] = pd.qcut(iris['petal_width'], q=3, labels=['small', 'medium', 'large'])
iris['petal_width_level'].value_counts().to_frame()
```

```python
pd.crosstab(iris['petal_width_level'], iris['class'])
```

```python
pd.crosstab(iris['petal_width_level'], iris['class'], normalize='all')
```
