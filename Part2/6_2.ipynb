```python
import pandas as pd
```

```python
product = pd.DataFrame({"상품코드": ["G1", "G2", "G3", "G4"], "상품명": ["노트북", "스마트폰", "태블릿"," 모니터"]})
orders = pd.DataFrame({"주문번호": [1001, 1002, 1003, 1004, 1005], "상품코드": ["G4", "G3", "G1", "G3", "G5"], "수량": [1, 4, 2, 2, 3]})
print(orders.merge(product, on="상품코드", how="inner"))
```

```python
orders.merge(product, on="상품코드", how="outer")
```

```python
orders.merge(product, on="상품코드", how="left")
```

```python
orders.merge(product, on="상품코드", how="right")
```
