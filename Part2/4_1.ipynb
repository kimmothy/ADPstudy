```python
import pandas as pd
students = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'], 'math': [85, None, 78, 90, 88], 'english': [90, None, 80, None, 92]})
students[students['name'] == 'Alice']
```

```python
print(students['name'] == 'Alice')
print(type(students['name'] == 'Alice'))
```

```python
students[(students['math'] >= 80) & (students['english'] > 80)]
```

```python
students.loc[6, 'name':'english'] = ['Frank', 82, 88]
students.loc[(students['english'] > 90) & (students['math'] > 85), 'P/F'] = 'Pass'
students.loc[students['P/F'] != 'Pass', 'P/F'] = 'Fail'
print(students)
```

```python
import numpy as np
conditions = [(students['math'] >= 90) & (students['english'] >= 90), (students['math'] >= 80) & (students['english'] >= 80)]
choice_list = ['Excellent', 'Good']
students['Grade'] = np.select(conditions, choice_list, default='Enhancing')
print(students)
```

```python
# 작동방식
print(conditions[0])
print(type(conditions[0]))
print(np.select(conditions, choice_list, default='Enhancing'))
```

```python
students['math'] = students['math'].fillna(students['math'].mean(numeric_only=True)) 
students['english'] = students['english'].fillna(0)
print(students)
```
