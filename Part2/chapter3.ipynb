```python
score = pd.DataFrame({"국어": [90, 80], "영어": [10, 10], "수학":[70, 70]}, index = ["장화", "홍련"])
score
```

```python
# score.loc["홍련"] = 100
score.loc["홍련", "영어"] = 100
# score["국어"] = score["국어"] - 5
score["국어"] -= 5
score
```

```python
print(score["국어"])
print(type(score["국어"]))
print(score["국어"]-5)
print(type(score["국어"]-5))
print(score.loc["홍련", "영어"])
print(type(score.loc["홍련", "영어"]))
```

```python
new_students = pd.DataFrame({"국어": [70, 85], "수학": [65, 100], "영어": [95, 65]}, index=["콩쥐", "팥쥐"])
score = pd.concat([score, new_students])

single_student = pd.Series({"국어": 100, "수학": 100, "영어":100}, name="햇님")

score.loc["햇님"] = single_student

score
```

```python
single_student = pd.Series({"국어": 100, "수학": 100, "영어":100})

score.loc["달님"] = single_student

score
```

```python
single_student = pd.Series({"국어": 100, "수학": 100, "영어":100}, name="무명")

score.loc["별님"] = single_student

score
```

```python
science = [80, 70, 90, 85, 75, 100, 95]
score["과학"] = science

score["학년"] = 1
score
```

```python
score["총점"] = score["국어"] + score["수학"] + score["영어"] + score["과학"]
score
# print(type(score["국어"] + score["수학"] + score["영어"] + score["과학"]))
```
