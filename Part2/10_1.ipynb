```python
import pandas as pd

landmark = pd.DataFrame(
    {
        "name": ["경복궁", "불국사", "첨성재", "광한루원", "창경궁"],
        "english_name": ["Gyeongbokgung Palace", "bulguksa", "cheomseongdae observatory", "gwanghallu", "Changgyeonggung Palace"],
        "location" : ["서울특별시 종로구 사직로 161", "경상북도 경주시 불국로 385", "경북 경주시 인왕동 839-1", "전라북도 남원시 요천로 1449", "서울특별시 종로구 창경궁로 185"],
    }
)
```

```python
landmark['name_upper'] = landmark['english_name'].str.upper()
landmark['name_lower'] = landmark['english_name'].str.lower()

landmark[['english_name','name_upper', 'name_lower']]
```

```python
landmark['name_capitalize'] = landmark['english_name'].str.capitalize()
landmark['name_title'] = landmark['english_name'].str.title()

landmark[['english_name','name_capitalize', 'name_title']]
```

```python
landmark['name_replace'] = landmark['english_name'].str.replace(" ", "_")

landmark['simple_locatoin'] = landmark['location'].str.replace("서울특별시", "서울").str.replace("경상북도", "경북").str.replace("전라북도", "전북").replace("경상북도", "경북")
landmark[['simple_locatoin', 'name_replace']]
```

```python
landmark["detailed address"] = landmark["location"].str.extract("(경주)")
landmark[["detailed address", "location"]]
```

```python
landmark["detailed address"] = landmark["location"].str.extract("(경주|서울)")
landmark[["detailed address", "location"]]
```

```python
landmark["detailed address"] = landmark["location"].str.extract("(\d+-?\d+)")
landmark[["detailed address", "location"]]
```

```python
landmark["is_palace"] = landmark["english_name"].str.contains("Palace", case=True)
landmark[["english_name", "is_palace"]]
```
